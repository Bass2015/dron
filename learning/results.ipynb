```python
import os
import torch
import numpy as np
import plotly.graph_objects as go

GAMMA = 0.99
LEARNING_RATE = 0.01
BATCH_SIZE = 64
ENTROPY_BETA = 0.01
NUM_ENVS = 50
REWARD_STEPS = 6 # How many steps ahead we take to approximate the total discounted reward for every action.
CLIP_GRAD = 0.1 # Prevents gradients from becoming too large. If L2 norm of gradients is bigger than this number, grad vector is clipped to this value.
ADAM_EPS = 1e-3

DEVICE = torch.device('mps')
def load_tensor(root, dir, tensor_name: str):
    file = os.path.join(root, dir, f'{tensor_name}.pt')
    with open(file, 'rb') as f:
        return torch.load(f)

root = 'debug'
max_probs_idx, l_v_idx, entropy_id = range(3)
epis = range(1,68)
losses = torch.vstack([load_tensor(root, f'Episode_{i}_data', 'losses') for i in epis])

mean, max = 0, 1
pol, val, ent = 0, 1, 2
max_entropy = -(np.tile(1/6, 6) * np.log(np.tile(1/6, 6))).sum()

fig = go.Figure(data=[
    go.Scatter(x=list(epis),
               y=losses[:,pol].flatten().numpy(),
               name="Policy Loss"),
    go.Scatter(x=list(epis),
               y=losses[:,val].flatten().numpy(),
               name="Value loss"),
    go.Scatter(x=list(epis),
               y=losses[:,ent].flatten().numpy(),
               name="Entropy"),
])
fig.add_hline(y=max_entropy)
fig.update_layout(width=800)
```
